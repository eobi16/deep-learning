```python
def get_technique_id(obj):

    refs = obj.get(
        "external_references",
        []
    )

    for ref in refs:

        if ref.get(
          "external_id"
        ):

            return ref[
             "external_id"
            ]

    return None
```

```python
technique_id_short = obj['ID'] # Get the short technique ID for display
technique_stix_id = obj['STIX ID'] # Get the STIX ID for cti methods

# Construct a STIX-compliant dictionary for the query
stix_object_for_relationships_query = {
    'id': technique_stix_id,
    'type': 'attack-pattern' # Assuming techniques are 'attack-pattern' STIX objects
}

# Get relationships for the current technique using the constructed STIX object
relationships = cti.get_relationships_by_object(stix_object_for_relationships_query)

mitigation_names = []
for rel in relationships:
    # Check if the relationship type is 'mitigates' and the source is the current technique
    if rel.get('relationship_type') == 'mitigates' and rel.get('source_ref') == technique_stix_id:
        mitigation_id = rel.get('target_ref')
        if mitigation_id:
            mitigation_obj = cti.get_object_by_attack_id(mitigation_id)
            if mitigation_obj and mitigation_obj.get('name'):
                mitigation_names.append(mitigation_obj['name'])

mitigations = list(set(mitigation_names)) # Remove duplicates

text = f"""

Technique ID: {technique_id_short}

Technique Name:
{obj.get('name')}

Description:
{description}

Tactics:
{tactics}

Platforms:
{platforms}

Detection:
{detection}

Mitigations:
{mitigations}

"""
```

```python
def retrieve_context(query):

    query_vector = model.encode(
        query
    )

    query_vector = np.array(
        query_vector
    ).reshape(1,-1)

    distances, indices = index.search(
        query_vector,
        5
    )

    contexts = []

    for idx in indices[0]:

        contexts.append(

            metadata[idx]

        )

    return contexts
```

### Resolving Gated Model Access

To access gated models like `meta-llama/Llama-3.1-8B-Instruct`, you must first:

1.  **Go to the model page on Hugging Face:** Visit [https://huggingface.co/meta-llama/Llama-3.1-8B-Instruct](https://huggingface.co/meta-llama/Llama-3.1-8B-Instruct).
2.  **Request Access:** If you don't have access, there will be a prominent button to request it. You might need to accept terms and conditions. Llama models often require an application.
3.  **Generate a Hugging Face Token:** Once granted access, go to your Hugging Face settings ([https://huggingface.co/settings/tokens](https://huggingface.co/settings/tokens)) and create a new **read access token**.
4.  **Store Token in Colab Secrets:** In Google Colab, click on the '🔑' (secrets) icon in the left sidebar. Add a new secret named `HF_TOKEN` and paste your Hugging Face token there. Make sure 'Notebook access' is enabled for this secret.

After completing these steps, the `huggingface_hub.login()` function should pick up the `HF_TOKEN` from your Colab secrets, or you can explicitly pass it. I've added a cell below to demonstrate loading the token from secrets.
